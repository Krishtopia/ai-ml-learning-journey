# PART II

let's see how do we create data from existing data

### 1. Column from series.

```python
seats_column = pd.Series([5]*500)
seats_column
```

```python
car_sales.head(10)
```

```python
car_sales["Seats"] = seats_column
```

```python
car_sales
```

```python
Tier_column = pd.Series([4]*6)
```

```python
Tier_column
```

```python
car_sales["Tier"] = Tier_column
```

```python
car_sales.head(10)
```

Let’s use the fillna() method to replace the missing (NaN) values in Tier section.

```python
car_sales["Tier"] = car_sales["Tier"].fillna(4)
```

```python
car_sales
```

### 2. Column from List.

```python
Fuel_economy = [7.5, 9.2, 5.0, 9.6, 8.7, 4.7, 8.8]
car_sales["Fuel per 100KM"] = Fuel_economy
```

HAHA - The ValueError occurs because our Python list contains only 7 values,

while car_sales has 500 rows. It's like this when creating a column using a list, its length must match the number of rows in the DataFrame.

```python
Fuel_economy = [ 8.8]*500
car_sales["Fuel per 100KM"] = Fuel_economy
```

```python
car_sales
```

```python
car_sales["Total fuel used"] = car_sales["Odometer (km)"]/100 * car_sales["Fuel per 100KM"]
```

```python
car_sales
```

### 3. Create a column from a single value

```python
car_sales["Roof"] = 1
```

```python
car_sales
```

```python
car_sales["Passed road safety"] = True
```

```python
car_sales.dtypes
```

```python
car_sales
```

Let’s drop (delete) one of the columns!

```python
car_sales = car_sales.drop("Total fuel used", axis=1)
```

```python
car_sales
```

```python
car_sales["Total fuel used"] = car_sales["Odometer (km)"]/100 * car_sales["Fuel per 100KM"]
```
