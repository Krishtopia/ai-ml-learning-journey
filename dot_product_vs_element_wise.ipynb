# Dot Product

```python
import numpy as np
```

```python
np.random.seed(0)

mat1 = np.random.randint(10, size=(5, 3))
mat2 = np.random.randint(10, size=(5, 3))
```

```python
mat1
```

```python
mat2
```

## Element-wise multiplication (Hadmard product)

```python
mat1 * mat2
```

## Dot Product

```python
np.dot(mat1, mat2)
```
