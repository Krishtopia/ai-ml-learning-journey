.iloc refers to position. remember, Python lists, Pandas DataFrames, & Series use 0-based positions, meaning the first element is at position 0.

.iloc refers to position whereas .loc refers to index label

```python
car_sales.iloc[5]
```

```python
car_sales.loc[5] 
```

Since both have the same index label and position, the outputs are the same.

# List Slicing with .iloc and .loc

```python
car_sales.head(10)
```

```python
car_sales.iloc[:4]
```

```python
car_sales.loc[:4]
```

Now, here's something interesting.

With .iloc, the ending position is excluded, just like normal Python list slicing. so, car_sales.iloc[:4] returns the first 4 rows.

But with .loc, the ending index label is included. so, car_sales.loc[:4] returns 5 rows because it includes the row with index label 4.

# Two Ways of Selecting a Column.

```python
car_sales["Make"] # The first way!
```

```python
car_sales["Colour"]
```

```python
car_sales.Make # The second way!
```

But. but. but. what if our column name contains spaces, like "Odometer (km)"? 

In that case, car_sales.Odometer (km) won't work coz dot notation doesn't support column names w/ spaces or special characters like parentheses.

Instead, we can use bracket notation: car_sales["Odometer (km)"] to access the column.
