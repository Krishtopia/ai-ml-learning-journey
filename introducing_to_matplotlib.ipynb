## First Method

```python
fig = plt.figure() # creates a figure
ax = fig.add_subplot() # adds some axes
plt.show()
```

# Second Method

```python
fig = plt.figure()
ax = fig.add_axes([1, 1, 1, 1])
ax.plot(x, y) # add some data
plt.show()
```

```python
fig = plt.figure()
ax = fig.add_axes([1, 1, 2, 1])
ax.plot(x, y)
```

```python
fig = plt.figure()
ax = fig.add_axes([0.1,0.1,0.8,0.8])
ax.plot(x, y)
plt.show()
```

## Third Method (recommended)

```python
fig, ax = plt.subplots()
ax.plot(x, [1, 2, 3, 9]);
```
