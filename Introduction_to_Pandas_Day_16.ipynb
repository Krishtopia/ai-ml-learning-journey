# IMPORTING DATA

```python
car_sales = pd.read_csv("cars_dataset.csv")
car_sales
```

# Importing Data

```python
car_sales.to_csv("exported-car-sales.csv", index = False)
```

Well, WE used index = False so pandas wouldn't write the DataFrame's index NO. as an additional column in  exported CSV file. as we only want our actual data columns, not the extra index column.
