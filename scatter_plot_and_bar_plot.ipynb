# Making Figures with Numpy Arrays

We want:

* Line plot
* Scatter plot
* Bar plot
* Histogram
* Subplots

```python
%matplotlib inline
import matplotlib.pyplot as plt
import numpy as np
```

## Create some data

```python
x = np.linspace(0, 10, 100)
```

```python
x[:10]
```

```python
x
```

## Plot the data and create a line plot

```python
# x = [1, 2, 3, 4]
```

```python
fig, ax = plt.subplots()
ax.plot(x, x**2);
```

## Using same data to make a scatter

```python
fig, ax = plt.subplots()
ax.scatter(x, np.exp(x))
```

```python
x, np.exp(x)
```

```python
fig, ax = plt.subplots()
ax.scatter(x, np.sin(x));
```

## Lets make a plot from dictionary

```python
nut_butter_prices = {
    "Almond butter": 10,
    "Peanut butter": 8,
    "Cashew butter": 12
}

fig, ax = plt.subplots()
ax.bar(x=nut_butter_prices.keys(), height=nut_butter_prices.values())
ax.set(title="Mine's Nut Butter Store",
      ylabel="Price($)",
      xlabel="Product name");
```
