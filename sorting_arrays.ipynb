# Sorting Arrays

```python
import numpy as np
```

```python
random_array = np.random.randint(10, size=(3, 5))
```

```python
random_array
```

```python
np.sort(random_array)
```

```python
np.argsort(random_array)
```

```python
a1 = np.array([1, 2, 3])
```

```python
a1
```

```python
np.sort(a1)
```

```python
np.argsort(a1)
```

```python
np.argmin(a1)
```

```python
np.argmin(random_array)
```

```python
np.argmax(a1)
```

```python
np.argmax(random_array)
```

```python
np.argmax(random_array, axis=0)
```
