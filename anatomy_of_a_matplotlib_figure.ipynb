# Matplotlib Example Workflow

Step 0: Import matplotlib and get it ready for plotting in Jupyter

```python
%matplotlib inline
import matplotlib.pyplot as plt
```

Step 1: Prepare Data

```python
x = [1, 2, 3, 4]
y = [11, 22, 33, 44]
```

Step 2: Setup Plot

```python
fig, ax = plt.subplots(figsize=(20,10))
```

Step 3: Plot the Data

```python
ax.plot(x, y)
```

Step 4: Customize plot

```python
ax.set(title="Simple Sowcase Plot",
      xlabel = "x-axis",
      ylabel = "y-axis")
```

Step 5: Save & show (you save the whole figure)

```python
fig.savefig("numpy-images/sample-plot.png")
```

```python
plt.show()
```

# Let's do this!!!!!!!

```python
%matplotlib inline
import matplotlib.pyplot as plt

x = [1, 2, 3, 4]
y = [11, 22, 33, 50]

fig, ax = plt.subplots(figsize=(10,10))
ax.plot(x, y)

ax.set(title="Simple Sowcase Plot",
      xlabel = "x-axis",
      ylabel = "y-axis")

plt.show()
```
