# Histograms and Subplots

```python
%matplotlib inline
import matplotlib.pyplot as plt
import numpy as np
```

```python
nut_butter_prices = {
    "Almond Butter": 10,
    "Peanut Butter": 8,
    "Cashew Butter": 12
}
```

```python
fig, ax = plt.subplots()
ax.bar(nut_butter_prices.keys(), nut_butter_prices.values())
```

```python
fig, ax = plt.subplots()
# ax.barh(list(nut_butter_prices.keys()), list(nut_butter_prices.values()))
ax.barh(nut_butter_prices.keys(), nut_butter_prices.values());
```

## Make some data for histograms and plot it

```python
x = np.random.randn(1000)
x
```

```python
fig, ax = plt.subplots()
ax.hist(x);
```

## Two options for subplots

### Subplot option 1

```python
fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(nrows=2,
                                             ncols=2,
                                             figsize=(20, 5))
# Plot to each different axis.
ax1.plot(x, x/2);

ax2.scatter(np.random.random(10), np.random.random(10));

ax3.bar(nut_butter_prices.keys(), nut_butter_prices.values());

ax4.hist(np.random.randn(1000)); #np.exp(x));
ax4.set(title="fuck you",
       ylabel="STFU");
```

## Subplot option 2

```python
fig, ax = plt.subplots(nrows=2,
                       ncols=2,
                       figsize=(20, 5)
                      )

# Plot to each different index
ax[0, 0].plot(x, x/2);
ax[0, 1].scatter(np.random.random(10), np.random.random(10));
ax[1, 0].hist(np.random.randn(1000));
ax[1, 1].bar(nut_butter_prices.keys(), nut_butter_prices.values())

ax[0, 0].set(title="fuck off");
```
